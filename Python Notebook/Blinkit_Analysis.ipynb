# Here I calculate a column Total Amount = price*quantity because it is important for revenue and creating dashboard

```python
df['Total Amount'] = df['Quantity']*df['Price']
```

# Calculate Column Order Hour = Hour(Order DateTime)



```python
df['Order Hour'] = df['Order DateTime'].dt.hour
```

```python
df['Day'] = df['Order DateTime'].dt.day_name()
```

```python
df.head()
```

```python
df['Month'] = df['Order DateTime'].dt.month_name()
```

```python
df['IsWeekend'] = df['Day'].isin(['Saturday', 'Sunday']).astype(int)
```

```python
df.head()
```

```python
def order_time(hour):
  if 5 <= hour < 12:
    return 'Morning'
  elif 12 <= hour < 17:
    return 'Afternoon'
  elif 17 <= hour < 21:
    return 'Evening'
  else:
    return 'Night'
```

```python
df['Order Period'] = df['Order Hour'].apply(order_time)
```

```python
df.head()
```

```python
df['High Order Value'] = df['Total Amount'] >= 500
```

```python
df.head(5)
```

```python
df['Fast Delivery'] = df['Delivery Time (mins)'] <= 30
```

```python
df.head(5)
```

```python
df.to_csv('new_blinkit.csv')
```

```python
df['IsWeekend'] = df['IsWeekend'].astype(str).str.replace('0', 'No').str.replace('1', 'Yes')
```

```python
df['High Order Value'] = df['High Order Value'].astype(str).str.replace('False', 'No').str.replace('True', 'Yes')
```

```python
df['Fast Delivery'] = df['Fast Delivery'].astype(str).str.replace('False', 'Slow').str.replace('True', 'Fast')
```
